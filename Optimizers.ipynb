```python
# What is gradient Descent :- Gradient Descent is an optimization technique thai is used in ml and dl to minimize the loss function,by 
# repeatedly adjusting the parameter in the direction to the steepest decrease.
```

```python
# What is learning rate :- learning rate is the hyperparameter that controls,how big the step a model take while updating its weight 
to reduce error during training.
```

```python
# Optimizers :- Optimizer is the algorithm that helps machine learning to improve the prediction by adjusting the parameter(Weight and Bias)
# to reduce error.
```

```python
In machine learning we have multiple types of optimizers like :-
(1).Batch Gradient Descent.
(2).Stochastic Gradient Descent.
(3).Mini Batch Gradient Descent.
(4).Momentum.
(5).Nesterov Accelerated Gradient(NAG)
```

```python
# Definitions:-
(1).Batch Gradient Descent:- Batch Gradient Descent uses entire training dataset to calculate the error(Gradient) and update the model
Weight once.
```

```python
(2).Stochastic Gradient Descent:- Instead of using all records in dataset, SGD uses just one training example to calculate the gradient and update the weights.
```

```python
(3).Mini-Batch Gradient Descent:- Mini-Batch Gradient Descent divides the training dataset into small group(batches) and updateds the model
weight after processing each group.
```

```python
(4).Momentum :- Momentum is an optimization technique that is uses the previous Gradient along with current Gradient to make the weight
update faster and smoother.
```

```python
(5).Nesterov Accelerated Gradient(NAG) :- NAG is an improved version of Momentum that look ahead at the future position
of the weight before calculating the gradient and updating them.
```

```python
(6).AdaGrad(Adaptive Gradient Algorithm):- AdaGrad is an optimizer that automatically adjust the learning rate for each parameter based it 
on previous(past) gradient.
```

```python
(7).RMSProp :- RMSProp is stands for root mean squared propogation is the optimizer that automatically adjust the learning rate for each parameter
by considering a moving average of recent squared gradient. that resolve the problem of vanishing gradient and gradient exploading.
```

```python
(8).Adam :- Adam is stands for Adaptive Moment Estimation is the optimization technique that combine the ideas of momentum and RMSProp 
to adjust the learning rate and update the model weight efficiently.
```
